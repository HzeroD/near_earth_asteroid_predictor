```python
import joblib
import random
import pandas as pd
import numpy as np


x_train_stats = pd.read_csv("../../artifacts/data/X_train_stats.csv").set_index('Unnamed: 0', drop=True)
x_train_stats.index.name = "stats"

x_train_no_moid_stats = joblib.load("../../artifacts/data/X_train_no_moid_stats.pkl")

x_train = joblib.load("../../artifacts/data/X_train.pkl")

x_train_no_moid = joblib.load("../../artifacts/data/X_train_no_moid.pkl")

def make_bad_data(cols, num, cat, stat_df):

    aberrant = pd.DataFrame([], columns=cols, index=range(num))
   
    for col in cols:
        if col in stat_df.columns.tolist():
            minimum = stat_df[col].loc["min"] 
            maximum = stat_df[col].loc["max"]

        for i in range(num):
            if col == 'condition_code':
                aberrant.loc[i, col] = int(np.random.choice([1, 8]))
                continue
            
            if col == 'size_category':
                aberrant.loc[i, col] = str(np.random.choice(["Small", "Medium"], 1)) 

            if col == "class_code":
                aberrant.loc[i, col] = 'Small' if col == cat[0] else 'APO'
                continue

            if col == "pha":
                aberrant.loc[i, col] = np.random.choice([True, False], 1)
                continue
            
            aberrant.loc[i, col] = round(float(random.uniform(minimum - (minimum * .2), maximum + (maximum * 0.2))),3)
            

    return aberrant
        
    
# cat = x_train.select_dtypes(include='str').columns.tolist()
# cols = x_train.columns.tolist()

cat = x_train_no_moid.select_dtypes(include=['str', 'category', 'bool']).columns.tolist()
cols = x_train_no_moid.columns.tolist()
print(cat)
df = make_bad_data(cols = cols, num=1000, cat=cat, stat_df = x_train_no_moid_stats)
x_train_no_moid
```

```python
df["pha"] = df["pha"].astype(int)
df["size_category"] = df["size_category"].astype(str)

df.to_dict(orient='records')

cat
```
